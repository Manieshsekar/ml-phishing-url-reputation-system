```python
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
DATA_PATH = "../data/raw/PhiUSIIL_Phishing_URL_Dataset.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
clean_df = df.copy()

print("Working copy created.")
print(clean_df.shape)
```

```python
missing = clean_df.isnull().sum()

missing_features = missing[missing > 0]

print("Columns containing missing values:")
print(missing_features)

print("\nTotal missing values:", missing.sum())
```

```python
duplicate_rows = clean_df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
url_column = "URL"

url_counts = clean_df[url_column].value_counts()

duplicate_urls = url_counts[url_counts > 1]

print("Number of URLs appearing more than once:",
      len(duplicate_urls))

print("\nTotal rows belonging to repeated URLs:",
      duplicate_urls.sum())
```

```python
TARGET_COLUMN = "label"

label_counts_per_url = (
    clean_df.groupby(url_column)[TARGET_COLUMN]
    .nunique()
)

conflicting_urls = label_counts_per_url[
    label_counts_per_url > 1
]

print("URLs with conflicting labels:",
      len(conflicting_urls))
```

```python
repeated_url_df = clean_df[
    clean_df[url_column].isin(duplicate_urls.index)
][[url_column, TARGET_COLUMN]]

print(repeated_url_df.head(20))
```

```python
print(clean_df[TARGET_COLUMN].value_counts(dropna=False))
print("Unique target values:")
print(sorted(clean_df[TARGET_COLUMN].unique()))
```

```python
print(clean_df[TARGET_COLUMN].dtype)
```

```python
nunique = clean_df.nunique()

constant_columns = nunique[nunique <= 1]

print("Constant columns:")
print(constant_columns)
```

```python
numeric_df = clean_df.select_dtypes(include=np.number)

dominance = {}

for column in numeric_df.columns:
    top_frequency = numeric_df[column].value_counts(
        normalize=True,
        dropna=False
    ).iloc[0]
    
    dominance[column] = top_frequency

dominance_series = pd.Series(dominance).sort_values(
    ascending=False
)

print(dominance_series.head(20))
```

```python
print(clean_df["FILENAME"].head(10))
print("\nUnique FILENAME values:",
      clean_df["FILENAME"].nunique())
print("Total rows:", len(clean_df))
print("Unique FILENAME:", clean_df["FILENAME"].nunique())
```

```python
object_columns = clean_df.select_dtypes(
    include=["object"]
).columns

print("Object columns:")
for column in object_columns:
    print("-", column)
for column in object_columns:
    print("\n" + "=" * 50)
    print(column)
    print(clean_df[column].head(5).tolist())
```

```python
numeric_columns = clean_df.select_dtypes(
    include=np.number
).columns

print("Number of numerical columns:",
      len(numeric_columns))

for column in numeric_columns:
    print("-", column)
```

```python
suspicious_columns = [
    column for column in clean_df.columns
    if any(
        word in column.lower()
        for word in ["label", "target", "class", "phish"]
    )
]

print("Potential target-related columns:")
print(suspicious_columns)
```

```python
correlations = (
    clean_df.select_dtypes(include=np.number)
    .corr()[TARGET_COLUMN]
    .sort_values(ascending=False)
)

print(correlations)
```

```python
feature_corr = (
    clean_df
    .select_dtypes(include=np.number)
    .drop(columns=[TARGET_COLUMN])
    .corr()
    .abs()
)

upper_triangle = feature_corr.where(
    np.triu(np.ones(feature_corr.shape), k=1).astype(bool)
)

high_corr_pairs = (
    upper_triangle
    .stack()
    .sort_values(ascending=False)
)

print(
    high_corr_pairs[
        high_corr_pairs > 0.95
    ].head(30)
)
```

```python
numeric_summary = clean_df.describe().T

print(numeric_summary)
```

```python
infinite_values = np.isinf(
    clean_df.select_dtypes(include=np.number)
).sum().sum()

print("Infinite numeric values:", infinite_values)
```

```python
memory_mb = (
    clean_df.memory_usage(deep=True).sum()
    / (1024 ** 2)
)

print(f"Memory usage: {memory_mb:.2f} MB")
```
